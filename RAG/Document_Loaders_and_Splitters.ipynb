### Documents
**Documents** in langChain are made up of a piece of text and optional metadata. The piece of text is what we interact with the language models.

### Document Loaders 
**Document Loaders** load the data from a source as documents.
Examples.
- CSVLoader
- PDFLoader
- WebBaseLoader
- JSONLoader
- YouTubeLoader
- WikiPediaLLoader

```python
from langchain.document_loaders import TextLoader

loader= TextLoader("sample.txt", encoding="utf-8")
document= loader.load()
print(document)
```

```python
document[0].page_content
```

### Transform Loader Example

```python
#CSVLoader
from langchain.document_loaders import CSVLoader

loader = CSVLoader(file_path="super_bowls.csv", encoding="utf-8")
documents = loader.load()

for document in documents:
    content= document.page_content
    metadata= document.metadata
    print(f"Content: {content}\nMetadata: {metadata}\n")
    print("------")
```

### Text Splitters
Once you've loaded documents, you will often want to tranform them to better suit your application. The simplest example is you want to split a long document into smaller chunks that can fit your model's context window. LangChain has a number of built-in document transformers that make it easy to split, combine, filter, and otherwise manipulate documents.

```python
from langchain.text_splitter import CharacterTextSplitter
text_splitter = CharacterTextSplitter(
    separator="\n\n",
    chunk_size=200, 
    chunk_overlap=20,
    length_function=len,
    is_separator_regex=False
    )
```

```python
loader = WebBaseLoader("https://www.ibm.com/")
data = loader.load()
```

```python
chunks= text_splitter.split_text(data[0].page_content)
len(chunks)
```

```python
for doc in chunks:
    print(doc)
    print("-----")  
```

```python
#RecursiveCharacterTextSplitter
from langchain.text_splitter import RecursiveCharacterTextSplitter
rectext_splitter = RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=20,
    length_function=len,
    is_separator_regex=False
)
```

```python
texts = rectext_splitter.create_documents([data[0].page_content])
```

```python
for text in texts:
    print(text)
    print("-----")  
```
