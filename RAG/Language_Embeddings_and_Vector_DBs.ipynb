### Language Embeddings
A **vector embedding** is a vector created as the numerical representation of typically non-numerical data objects such as NL, Image, Sound or Video.
- Embeddings capture the inherent properties and relationships of the original data in a numerical vector format and are often used in Machine Learning use cases.
- In case of NLP, embedding vectors capture the **semantic and syntactic senses**
- The transformation process from the original data is performed by what's known as an "embedding model" and involves complex ML techniques including sophisticated Neural Networks

### Vector Databases
Vector databases provide the ability to store and retrieve **Embedding Vector** as high-dimensional points. They add capabilities for efficient and fast lookup of **nearest-neighbors in the N-dimensional space**. They are typically powered by k-nearest neighbor (k-NN) indexes. Vector databases provide additional capabilities like data management, fault tolerence, authentication and access control, and a query engine.
